### 1.1 Preprocessing Steps

In this project, the preprocessing pipeline is intentionally **multi-stage** and designed to preserve linguistic features relevant to emotion. The steps used in the actual code are:

#### **Data Loading & Merging**

* Loaded **final_posts.json**, **data_identification.csv**, and **emotion.csv**.
* Extracted:

  * `post_id`
  * `text`
  * `hashtags`
  * `num_hashtags`
* Merged them into a unified dataframe, then split into **train** and **test** using the existing `split` field.

#### **Text Cleaning**

Each post goes through `clean_text()`, which:

* Lowercases all text
* Removes:

  * URLs
  * emails
  * mentions e.g., `@username`
  * hashtag symbols (but keeps the word)
* Collapses repeated punctuation (`!!!!` → `!`)
* Removes special characters except `. , ? !`
* Normalizes whitespace

#### **Advanced Preprocessing**

The step `advanced_preprocess()` performs high-quality NLP preprocessing:

* **Tokenization** using NLTK
* Removes normal stopwords **but retains emotion-important words**, such as:

  * `not`, `never`, `very`, `too`, `hate`, `love`, `feel`, etc.
* **Lemmatization**
* Removes tokens shorter than 2 characters

This makes text cleaner while preserving emotional signals.

---

### 1.2 Feature Engineering Steps

The project uses **three categories of features**:

1. **Linguistic features**
2. **Text-vectorization features**
3. **POS-tag features**

#### **Linguistic & Statistical Features**

The function `extract_linguistic_features()` extracts **35+ numeric features**, including:

##### **Text statistics**

* text length
* word count
* sentence count
* average sentence length
* average word length
* unique word ratio

##### **Punctuation features**

* exclamation count & ratio
* question mark count & ratio
* uppercase count & ratio
* period/comma counts

##### **Emotion lexicon word counts**

Manually constructed lexicons:

* anger words
* disgust words
* fear words
* sadness words
* joy words
* surprise words

Counts the occurrence of each category inside the text.

##### **Sentiment features (TextBlob)**

* polarity
* subjectivity

##### **Negation features**

* negation_count
* negation_ratio

##### **Intensifier features**

* very, really, incredibly, totally, etc.

##### **Hashtag features**

* num_hashtags
* has_hashtags

##### **Special pattern features**

* has_ellipsis
* repeated_chars


#### **TF-IDF & Count Vectorizer Features**

The model uses **multiple text representations**, all created in `create_multiple_text_features()`:

| Feature Type                 | Settings                            | Size |
| ---------------------------- | ----------------------------------- | ---- |
| TF-IDF (word-grams)          | n-grams 1–3, max_features=3000      | 3000 |
| TF-IDF (char-grams)          | char n-grams 2–5, max_features=2000 | 2000 |
| Bag-of-Words CountVectorizer | n-grams 1–2, max_features=2000      | 2000 |

These are used differently depending on model:

* **Naive Bayes uses CountVectorizer**
* **All other models use TF-IDF (word + char) + numerical features**


#### **POS Tag Features**

Extracts:

* noun_ratio
* verb_ratio
* adjective_ratio
* adverb_ratio

These ratios help capture syntactic structure correlated with emotions.


#### **Scaling & Combination**

* Numerical features are standardized using **StandardScaler**
* All features are combined via `hstack()`:

  ```
  [TF-IDF word] + [TF-IDF char] + [scaled numeric features]
  ```

## 1. Preprocessing Steps

````python
### Add the code related to the preprocessing steps in cells inside this section
def load_data():
    """Load all datasets"""
    print("Loading datasets...")
    
    # Load JSON data
    with open('final_posts.json', 'r', encoding='utf-8') as f:
        posts_data = json.load(f)
    
    # Extract post information from JSON
    posts_list = []
    for item in posts_data:
        post_id = item['root']['_source']['post']['post_id']
        text = item['root']['_source']['post']['text']
        hashtags = item['root']['_source']['post']['hashtags']
        posts_list.append({
            'id': post_id,
            'text': text,
            'hashtags': hashtags,
            'num_hashtags': len(hashtags)
        })
    
    posts_df = pd.DataFrame(posts_list)
    
    # Load other CSV files
    identification_df = pd.read_csv('data_identification.csv')
    emotion_df = pd.read_csv('emotion.csv')
    
    print(f"  - Posts loaded: {len(posts_df)}")
    print(f"  - Identifications loaded: {len(identification_df)}")
    print(f"  - Emotions loaded: {len(emotion_df)}")
    
    return posts_df, identification_df, emotion_df

def clean_text(text):
    """Clean and normalize text"""
    if pd.isna(text):
        return ""
    
    # Convert to lowercase
    text = text.lower()
    
    # Remove URLs
    text = re.sub(r'http\S+|www\S+|https\S+', '', text, flags=re.MULTILINE)
    
    # Remove email addresses
    text = re.sub(r'\S+@\S+', '', text)
    
    # Remove mentions (but keep hashtag content)
    text = re.sub(r'@\w+', '', text)
    text = re.sub(r'#', '', text)
    
    # Replace multiple punctuation with single
    text = re.sub(r'([!?.]){2,}', r'\1', text)
    
    # Remove special characters but keep important punctuation
    text = re.sub(r'[^a-z\s!?.,]', ' ', text)
    
    # Remove extra whitespace
    text = re.sub(r'\s+', ' ', text).strip()
    
    return text

def advanced_preprocess(text):
    """Advanced text preprocessing with emotion preservation"""
    
    # Clean text
    text = clean_text(text)
    
    # Tokenize
    tokens = word_tokenize(text)
    
    # Remove stopwords but keep emotion-related ones
    stop_words = set(stopwords.words('english'))
    # Preserve important words for emotion detection
    keep_words = {'not', 'no', 'never', 'very', 'too', 'so', 'but', 'however', 
                  'yet', 'although', 'though', 'really', 'quite', 'just', 'only',
                  'hate', 'love', 'like', 'want', 'need', 'feel', 'think'}
    stop_words = stop_words - keep_words
    tokens = [t for t in tokens if t not in stop_words or t in keep_words]
    
    # Lemmatization
    lemmatizer = WordNetLemmatizer()
    tokens = [lemmatizer.lemmatize(t) for t in tokens]
    
    # Remove very short tokens
    tokens = [t for t in tokens if len(t) > 1]
    
    return ' '.join(tokens)

def merge_and_prepare_data(posts_df, identification_df, emotion_df):
    """Merge all dataframes and prepare train/test sets"""
    print("\nMerging and preparing data...")
    
    # Merge posts with identification
    df = posts_df.merge(identification_df, on='id', how='left')
    
    # Merge with emotion labels
    df = df.merge(emotion_df, on='id', how='left')
    
    # Apply text preprocessing
    print("Preprocessing text data...")
    df['clean_text'] = df['text'].apply(lambda x: advanced_preprocess(x))
    
    # Split into train and test
    train_df = df[df['split'] == 'train'].copy()
    test_df = df[df['split'] == 'test'].copy()
    
    print(f"  - Training samples: {len(train_df)}")
    print(f"  - Test samples: {len(test_df)}")
    print(f"  - Emotion distribution in training set:")
    print(train_df['emotion'].value_counts())
    
    return train_df, test_df, df

# Load and preprocess data
posts_df, identification_df, emotion_df = load_data()
train_df, test_df, full_df = merge_and_prepare_data(posts_df, identification_df, emotion_df)
train_df = train_df.reset_index(drop=True)
test_df  = test_df.reset_index(drop=True)
````

## 2. Feature Engineering Steps

````python
### Add the code related to the feature engineering steps in cells inside this section
def extract_linguistic_features(df):
    """Extract advanced linguistic and emotion-specific features"""
    print("Extracting advanced linguistic features...")
    
    features = pd.DataFrame()
    
    # === BASIC TEXT STATISTICS ===
    features['text_length'] = df['text'].apply(len)
    features['word_count'] = df['text'].apply(lambda x: len(str(x).split()))
    features['clean_word_count'] = df['clean_text'].apply(lambda x: len(str(x).split()))
    features['avg_word_length'] = df['text'].apply(
        lambda x: np.mean([len(word) for word in str(x).split()]) if len(str(x).split()) > 0 else 0
    )
    features['sentence_count'] = df['text'].apply(lambda x: len(sent_tokenize(str(x))))
    features['avg_sentence_length'] = features['word_count'] / (features['sentence_count'] + 1)
    
    # === PUNCTUATION FEATURES ===
    features['exclamation_count'] = df['text'].apply(lambda x: str(x).count('!'))
    features['question_count'] = df['text'].apply(lambda x: str(x).count('?'))
    features['period_count'] = df['text'].apply(lambda x: str(x).count('.'))
    features['comma_count'] = df['text'].apply(lambda x: str(x).count(','))
    features['uppercase_count'] = df['text'].apply(lambda x: sum(1 for c in str(x) if c.isupper()))
    features['uppercase_ratio'] = features['uppercase_count'] / (features['text_length'] + 1)
    
    # Punctuation ratios
    features['exclamation_ratio'] = features['exclamation_count'] / (features['word_count'] + 1)
    features['question_ratio'] = features['question_count'] / (features['word_count'] + 1)
    
    # === EMOTION-SPECIFIC LEXICONS ===
    # Anger words
    anger_words = {'angry', 'anger', 'rage', 'furious', 'mad', 'irritated', 'annoyed', 
                   'frustrated', 'hate', 'hatred', 'outrage', 'pissed', 'damn'}
    
    # Disgust words
    disgust_words = {'disgust', 'disgusting', 'gross', 'nasty', 'sick', 'revolting',
                     'repulsive', 'awful', 'terrible', 'horrible', 'yuck'}
    
    # Fear words
    fear_words = {'fear', 'scared', 'afraid', 'terrified', 'anxious', 'worry', 'nervous',
                  'panic', 'dread', 'frightened', 'alarmed', 'horror', 'terror'}
    
    # Sadness words
    sadness_words = {'sad', 'sadness', 'depressed', 'unhappy', 'miserable', 'sorrow',
                     'grief', 'heartbroken', 'lonely', 'disappointed', 'cry', 'tears'}
    
    # Joy words
    joy_words = {'happy', 'joy', 'excited', 'great', 'wonderful', 'amazing', 'excellent',
                 'love', 'glad', 'delighted', 'cheerful', 'thrilled', 'fantastic', 'awesome'}
    
    # Surprise words
    surprise_words = {'surprise', 'surprised', 'shocked', 'unexpected', 'sudden', 'wow',
                      'unbelievable', 'astonished', 'amazed', 'stunned'}
    
    # Count emotion words
    features['anger_word_count'] = df['clean_text'].apply(
        lambda x: sum(1 for word in str(x).split() if word in anger_words)
    )
    features['disgust_word_count'] = df['clean_text'].apply(
        lambda x: sum(1 for word in str(x).split() if word in disgust_words)
    )
    features['fear_word_count'] = df['clean_text'].apply(
        lambda x: sum(1 for word in str(x).split() if word in fear_words)
    )
    features['sadness_word_count'] = df['clean_text'].apply(
        lambda x: sum(1 for word in str(x).split() if word in sadness_words)
    )
    features['joy_word_count'] = df['clean_text'].apply(
        lambda x: sum(1 for word in str(x).split() if word in joy_words)
    )
    features['surprise_word_count'] = df['clean_text'].apply(
        lambda x: sum(1 for word in str(x).split() if word in surprise_words)
    )
    
    # === SENTIMENT FEATURES (using TextBlob) ===
    print("Computing sentiment scores...")
    sentiments = df['text'].apply(lambda x: TextBlob(str(x)).sentiment)
    features['polarity'] = sentiments.apply(lambda x: x.polarity)
    features['subjectivity'] = sentiments.apply(lambda x: x.subjectivity)
    
    # === NEGATION FEATURES ===
    negation_words = {'not', 'no', 'never', 'neither', 'nobody', 'nothing', 'nowhere', 
                     "n't", 'barely', 'hardly', 'scarcely'}
    features['negation_count'] = df['clean_text'].apply(
        lambda x: sum(1 for word in str(x).split() if word in negation_words)
    )
    features['negation_ratio'] = features['negation_count'] / (features['word_count'] + 1)
    
    # === INTENSIFIER FEATURES ===
    intensifiers = {'very', 'really', 'so', 'too', 'extremely', 'absolutely', 
                   'completely', 'totally', 'quite', 'incredibly'}
    features['intensifier_count'] = df['clean_text'].apply(
        lambda x: sum(1 for word in str(x).split() if word in intensifiers)
    )
    
    # === HASHTAG FEATURES ===
    features['num_hashtags'] = df['num_hashtags']
    features['has_hashtags'] = (features['num_hashtags'] > 0).astype(int)
    
    # === SPECIAL PATTERNS ===
    features['has_ellipsis'] = df['text'].apply(lambda x: 1 if '...' in str(x) else 0)
    features['repeated_chars'] = df['text'].apply(
        lambda x: len(re.findall(r'(.)\1{2,}', str(x)))
    )
    
    # === UNIQUE WORD RATIO ===
    features['unique_word_ratio'] = df['clean_text'].apply(
        lambda x: len(set(str(x).split())) / (len(str(x).split()) + 1)
    )
    
    print(f"  - Extracted {features.shape[1]} linguistic features")
    
    return features

def create_multiple_text_features(train_texts, test_texts):
    """Create multiple types of text vectorization features"""
    print("\nCreating multiple text representation features...")
    
    # === TF-IDF with word n-grams ===
    print("  - Creating TF-IDF word features...")
    tfidf_word = TfidfVectorizer(
        max_features=3000,
        ngram_range=(1, 3),
        min_df=2,
        max_df=0.9,
        sublinear_tf=True,
        strip_accents='unicode'
    )
    X_train_tfidf_word = tfidf_word.fit_transform(train_texts)
    X_test_tfidf_word = tfidf_word.transform(test_texts)
    
    # === TF-IDF with character n-grams ===
    print("  - Creating TF-IDF character features...")
    tfidf_char = TfidfVectorizer(
        max_features=2000,
        analyzer='char',
        ngram_range=(2, 5),
        min_df=2,
        max_df=0.9,
        sublinear_tf=True
    )
    X_train_tfidf_char = tfidf_char.fit_transform(train_texts)
    X_test_tfidf_char = tfidf_char.transform(test_texts)
    
    # === Count Vectorizer (for Naive Bayes) ===
    print("  - Creating Count features...")
    count_vec = CountVectorizer(
        max_features=2000,
        ngram_range=(1, 2),
        min_df=2,
        max_df=0.9
    )
    X_train_count = count_vec.fit_transform(train_texts)
    X_test_count = count_vec.transform(test_texts)
    
    print(f"  - TF-IDF word shape: {X_train_tfidf_word.shape}")
    print(f"  - TF-IDF char shape: {X_train_tfidf_char.shape}")
    print(f"  - Count shape: {X_train_count.shape}")
    
    return (X_train_tfidf_word, X_test_tfidf_word, 
            X_train_tfidf_char, X_test_tfidf_char,
            X_train_count, X_test_count,
            tfidf_word, tfidf_char, count_vec)

def create_pos_tag_features(df):
    """Create features based on POS tags"""
    print("Creating POS tag features...")
    
    def get_pos_counts(text):
        tokens = word_tokenize(str(text).lower())
        pos_tags = nltk.pos_tag(tokens)
        
        # Count different POS types
        noun_count = sum(1 for _, pos in pos_tags if pos.startswith('NN'))
        verb_count = sum(1 for _, pos in pos_tags if pos.startswith('VB'))
        adj_count = sum(1 for _, pos in pos_tags if pos.startswith('JJ'))
        adv_count = sum(1 for _, pos in pos_tags if pos.startswith('RB'))
        
        total = len(pos_tags) + 1
        return noun_count/total, verb_count/total, adj_count/total, adv_count/total
    
    pos_features = df['text'].apply(get_pos_counts)
    
    pos_df = pd.DataFrame(
        pos_features.tolist(),
        columns=['noun_ratio', 'verb_ratio', 'adj_ratio', 'adv_ratio']
    )
    
    return pos_df

# Extract all features
train_features = extract_linguistic_features(train_df)
test_features = extract_linguistic_features(test_df)

train_features = train_features.reset_index(drop=True)
test_features  = test_features.reset_index(drop=True)

# Add POS features
train_pos = create_pos_tag_features(train_df)
test_pos = create_pos_tag_features(test_df)
train_features = pd.concat([train_features, train_pos], axis=1)
test_features = pd.concat([test_features, test_pos], axis=1)

# Create text features
(X_train_tfidf_word, X_test_tfidf_word,
 X_train_tfidf_char, X_test_tfidf_char,
 X_train_count, X_test_count,
 tfidf_word, tfidf_char, count_vec) = create_multiple_text_features(
    train_df['clean_text'], 
    test_df['clean_text']
)

# Scale numerical features
print("\nScaling numerical features...")
scaler = StandardScaler()
train_features_scaled = scaler.fit_transform(train_features)
test_features_scaled = scaler.transform(test_features)

# Combine all features
from scipy.sparse import hstack, csr_matrix

X_train_combined = hstack([
    X_train_tfidf_word,
    X_train_tfidf_char,
    csr_matrix(train_features_scaled)
])

X_test_combined = hstack([
    X_test_tfidf_word,
    X_test_tfidf_char,
    csr_matrix(test_features_scaled)
])

print(f"\nFinal combined feature matrix shape:")
print(f"  - Training: {X_train_combined.shape}")
print(f"  - Test: {X_test_combined.shape}")

# Encode labels
label_encoder = LabelEncoder()
y_train = label_encoder.fit_transform(train_df['emotion'])
print(f"\nLabel encoding: {dict(zip(label_encoder.classes_, range(len(label_encoder.classes_))))}")
````

## 3. Model Implementation Steps

````python
### Add the code related to the model implementation steps in cells inside this section
def train_multiple_models(X_train, y_train, X_train_count):
    """Train multiple models with different configurations"""
    print("Training multiple advanced models...\n")
    
    models = {
        'Logistic Regression (L2)': LogisticRegression(
            max_iter=1000, C=2.0, solver='saga', random_state=42, n_jobs=-1
        ),
        'Linear SVM': LinearSVC(
            C=0.5, max_iter=2000, random_state=42, dual=False
        ),
        'Naive Bayes (Count)': MultinomialNB(alpha=0.1),
        'Random Forest': RandomForestClassifier(
            n_estimators=300, max_depth=50, min_samples_split=5,
            min_samples_leaf=2, random_state=42, n_jobs=-1
        ),
        'Gradient Boosting': GradientBoostingClassifier(
            n_estimators=200, max_depth=7, learning_rate=0.1,
            subsample=0.8, random_state=42
        )
    }
    
    results = {}
    
    for name, model in models.items():
        print(f"Training {name}...")
        
        # Use count features for Naive Bayes, combined for others
        if 'Naive Bayes' in name:
            X_train_model = X_train_count
        else:
            X_train_model = X_train
        
        # Cross-validation
        cv_scores = cross_val_score(
            model, X_train_model, y_train, 
            cv=5, scoring='accuracy', n_jobs=-1
        )
        
        # Train on full training set
        model.fit(X_train_model, y_train)
        
        # Store results
        results[name] = {
            'model': model,
            'cv_mean': cv_scores.mean(),
            'cv_std': cv_scores.std(),
            'use_count': 'Naive Bayes' in name
        }
        
        print(f"  - CV Accuracy: {cv_scores.mean():.4f} (+/- {cv_scores.std():.4f})")
        print()
    
    return results

def create_ensemble(results, X_train, y_train, X_train_count):
    """Create an ensemble of top models"""
    print("Creating ensemble model...")
    
    # Select top 3 models
    sorted_models = sorted(results.items(), key=lambda x: x[1]['cv_mean'], reverse=True)
    top_models = sorted_models[:3]
    
    print("Top 3 models for ensemble:")
    for name, info in top_models:
        print(f"  - {name}: {info['cv_mean']:.4f}")
    
    # Note: VotingClassifier with sparse matrices requires all models to use same input
    # So we'll just use the best single model
    best_model_name = sorted_models[0][0]
    best_model = sorted_models[0][1]['model']
    
    print(f"\nBest single model: {best_model_name}")
    print(f"CV Accuracy: {sorted_models[0][1]['cv_mean']:.4f}")
    
    return best_model, best_model_name, sorted_models[0][1]['use_count']

def make_predictions(model, X_test, X_test_count, test_df, label_encoder, use_count=False):
    """Make predictions on test set"""
    print("\n\nMaking predictions on test set...")
    
    # Use appropriate feature set
    X_test_model = X_test_count if use_count else X_test
    
    # Predict
    y_pred = model.predict(X_test_model)
    
    # Decode labels
    predicted_emotions = label_encoder.inverse_transform(y_pred)
    
    # Create submission dataframe
    submission_df = pd.DataFrame({
        'id': test_df['id'].values,
        'emotion': predicted_emotions
    })
    
    print(f"Predictions made for {len(submission_df)} test samples")
    print("\nPredicted emotion distribution:")
    print(submission_df['emotion'].value_counts())
    
    return submission_df

# Train all models
all_results = train_multiple_models(X_train_combined, y_train, X_train_count)

# Create ensemble / select best model
best_model, best_model_name, use_count = create_ensemble(
    all_results, X_train_combined, y_train, X_train_count
)

# Make predictions
submission_df = make_predictions(
    best_model, X_test_combined, X_test_count, 
    test_df, label_encoder, use_count
)

# Save submission file
submission_df.to_csv('submission.csv', index=False)
print("\n✓ Submission file saved as 'submission.csv'")
````
